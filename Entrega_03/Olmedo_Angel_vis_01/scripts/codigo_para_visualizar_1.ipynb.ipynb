```python
import pandas as pd
import altair as alt
from google.colab import files

print("Por favor, selecciona tu archivo 'Olmedo_Angel_basededatoslimpia.csv':")
uploaded = files.upload()

# Obtenemos el nombre del archivo que acabas de subir
nombre_archivo = list(uploaded.keys())[0]

# 1. Cargar la base de datos
df = pd.read_csv(nombre_archivo)

# Asegurar formato numérico correcto
df['GOLES CHILE'] = pd.to_numeric(df['GOLES CHILE'])
df['GOLES RIVAL'] = pd.to_numeric(df['GOLES RIVAL'])

# 2. Crear gráfico de barras con Altair
chart = alt.Chart(df).mark_bar().encode(
    x=alt.X('GENERACION:N', title='Generación de la Selección', sort=['Generación 1990-1997', 'Generación 1998-2007', 'Generación Dorada']),
    y=alt.Y('mean(GOLES CHILE):Q', title='Promedio de Goles Anotados'),
    color=alt.Color('TIPO:N', title='Tipo de Partido', scale=alt.Scale(scheme='category10')),
    column=alt.Column('TIPO:N', title='Rendimiento: Oficiales vs Amistosos'),
    tooltip=['GENERACION', 'TIPO', alt.Tooltip('mean(GOLES CHILE):Q', format='.2f')]
).properties(
    width=220,
    height=300,
    title='Rendimiento de La Roja por Generación'
).interactive()

# Mostrar el gráfico en Colab
display(chart)

# 3. Guardar automáticamente como HTML para la entrega
chart.save('vis_01.html')
print("¡Listo! Archivo vis_01.html guardado con éxito.")
```
