```python
import pandas as pd
import altair as alt

# 1. Cargar la base de datos con el nombre corto
df = pd.read_csv('datos.csv', encoding='latin1', sep=';')

# 2. Renombrar columnas para evitar caracteres especiales
df.columns = ['num_partido', 'ano', 'Generacion', 'Rival', 'Condicion', 'Tipo',
              'Competencia', 'Goles_Chile', 'Goles_Rival', 'Dif_Goles', 'Resultado', 'Ciudad']

# 3. Limpiar espacios extra
df['Generacion'] = df['Generacion'].str.strip()
df['Tipo'] = df['Tipo'].str.strip()

# 4. Crear el gráfico interactivo de Altair
chart = alt.Chart(df).mark_bar().encode(
    x=alt.X('Generacion:N', title='Generación', sort=['Generación 1990-1997', "Generación Francia '98", 'Generación Dorada', 'Post Generación Dorada']),
    y=alt.Y('mean(Goles_Chile):Q', title='Promedio de Goles de Chile'),
    color=alt.Color('Tipo:N', title='Tipo de Partido'),
    column=alt.Column('Tipo:N', title='Condición Reglamentaria')
).properties(
    title='Promedio de Goles de La Roja por Generación y Tipo de Partido',
    width=180,
    height=280
)

# 5. Guardar el gráfico interactivo en HTML y mostrarlo
chart.save('vis_01.html')
chart
```
